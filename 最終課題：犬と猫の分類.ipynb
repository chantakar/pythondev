```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# データのパス
base_dir = r"C:\Users\chant_s2mci8i\Downloads\pythondev\dog_cat_photos"
train_dir = base_dir + r"\train"
test_dir = base_dir + r"\test"

# MobileNetV2 の入力サイズ（デフォルトは 224x224）
img_size = (224, 224)
batch_size = 32

# データ水増し（データ拡張）
train_datagen = ImageDataGenerator(
    rescale=1./255,           # 正規化
    rotation_range=20,        # 回転
    width_shift_range=0.2,    # 横移動
    height_shift_range=0.2,   # 縦移動
    shear_range=0.2,          # せん断
    zoom_range=0.2,           # ズーム
    horizontal_flip=True,     # 左右反転
    fill_mode='nearest'
)

test_datagen = ImageDataGenerator(rescale=1./255)

# データセットを生成
train_generator = train_datagen.flow_from_directory(
    train_dir,
    target_size=img_size,
    batch_size=batch_size,
    class_mode='binary'
)

test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=img_size,
    batch_size=batch_size,
    class_mode='binary'
)

# MobileNetV2 のベースモデルをロード（事前学習済み）
base_model = keras.applications.MobileNetV2(
    input_shape=img_size + (3,),
    include_top=False,   # 最後の全結合層を外す
    weights='imagenet'
)

# 転移学習：ベースモデルの重みを固定
base_model.trainable = False

# モデル構築
model = keras.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(1, activation='sigmoid')  # 犬/猫の二値分類
])

# コンパイル
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# 学習
history = model.fit(
    train_generator,
    epochs=10,
    validation_data=test_generator
)

# 評価
loss, acc = model.evaluate(test_generator)
print(f"Test Accuracy: {acc:.4f}")
```
