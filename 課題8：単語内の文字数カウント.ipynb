```python
# 英単語を格納する空のリスト
words = []

# アルファベット小文字の各文字をキー、値を0とする辞書
char_count = {chr(i): 0 for i in range(ord('a'), ord('z') + 1)}

# 英単語の入力受付
while True:
    word = input("英単語を入力してください：").strip().lower()
    if word == "":
        break
    words.append(word)
    print(f"入力された英単語：{word}")  # 入力ごとに表示
    for c in word:
        if c in char_count:
            char_count[c] += 1

# 入力された英単語リスト（ABC順）
words_sorted = sorted(words)
print(f"入力した英単語：{words_sorted}")

# 1文字以上あった文字の個数を表示
for k, v in char_count.items():
    if v > 0:
        print(f"{k}が{v}個ありました")
```
